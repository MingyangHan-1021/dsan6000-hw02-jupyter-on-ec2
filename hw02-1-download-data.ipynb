```python
# Setup: imports and constants
import os
import boto3

SRC_BUCKET = "dsan6000-wikipedia"
PREFIX = "hourly_parquet/"
LOCAL_DIR = "data"
```

```python
# Connect to S3 and list everything under the prefix
s3 = boto3.client("s3")
response = s3.list_objects_v2(Bucket=SRC_BUCKET, Prefix=PREFIX)

# The raw list of objects returned by S3
contents = response["Contents"]

# Keep only the real files; the folder placeholder has size 0
keys = []

for obj in contents:
    key = obj["Key"]
    size = obj["Size"]

    if size > 0:
        keys.append(key)

print(f"Total {len(keys)} parquet files were discovered")   # should be 24
```

```python
# Make sure the local folder exists
os.makedirs(LOCAL_DIR, exist_ok=True)

for key in keys:
    filename = os.path.basename(key)
    dest = os.path.join(LOCAL_DIR, filename)

    s3.download_file(Bucket=SRC_BUCKET, Key=key, Filename=dest)

    print(f"Downloaded {filename}")
```
