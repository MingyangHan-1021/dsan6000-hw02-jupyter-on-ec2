```python
# imports libraries
import os
import boto3

LOCAL_DIR = "data"
MY_BUCKET = "dsan6000-mh2393"
DEST_PREFIX = "wikipedia-hourly/"

s3 = boto3.client("s3")
```

```python
# new S3 bucket
s3.create_bucket(Bucket=MY_BUCKET)

print(f"Created bucket {MY_BUCKET}")
```

```python
# Upload parquet files to new S3 bucket
filenames = os.listdir(LOCAL_DIR)

for filename in filenames:
    local_path = os.path.join(LOCAL_DIR, filename)
    dest_key = DEST_PREFIX + filename

    s3.upload_file(Filename=local_path, Bucket=MY_BUCKET, Key=dest_key)

    print(f"Uploaded {filename}")
```

```python
# Confirm
response = s3.list_objects_v2(Bucket=MY_BUCKET, Prefix=DEST_PREFIX)

print(f"{len(response['Contents'])} files were found")   # should be 24
```
